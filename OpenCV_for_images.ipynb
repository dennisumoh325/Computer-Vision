```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# reading the image
image = cv2.imread(r'C:\\Users\\New\\Desktop\\Jupyter Notebook\\ML\\Computer Vision\\OpenCV for images\\opencv.jpg')

#displaying the image with opencv
cv2.imshow('Image', image)
plt.imshow(image)
```

```python


# Check whether the image loaded
if image is not None:

    # Convert BGR to RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Display image
    plt.imshow(image_rgb)
    plt.title("My Image")
    plt.axis("off")
    plt.show()

else:
    print("Image not found. Check your file path.")
```
