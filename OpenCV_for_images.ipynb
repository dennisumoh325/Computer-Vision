```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# reading the image
image = cv2.imread(r'C:\\Users\\New\\Desktop\\Jupyter Notebook\\ML\\Computer Vision\\OpenCV for images\\opencv.jpg')

#displaying the image with opencv
cv2.imshow('Image', image)
plt.imshow(image)
```

```python


# Check whether the image loaded
if image is not None:

    # Convert BGR to RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Display image
    plt.imshow(image_rgb)
    plt.title("My Image")
    plt.axis("off")
    plt.show()

else:
    print("Image not found. Check your file path.")
```

### Resizing the image

```python
height = 400
width = 400
dimensions = (width, height)
new_image = cv2.resize(image, dimensions )
print("Original image shape:", image.shape)
print("Resized image shape:", new_image.shape)
cv2.imshow('Resized Image', new_image)
plt.imshow(new_image)
```

```python
# Resizing the image using INTER_AREA interpolation
height = 400
width = 400
dimensions = (width, height)
new_image = cv2.resize(image, dimensions, interpolation=cv2.INTER_AREA) 
print("Original image shape:", image.shape)
print("Resized image shape:", new_image.shape)
cv2.imshow('Resized Image', new_image)
plt.imshow(new_image)
```

```python
# Resizing with scaling factors
height = image.shape[0]
width = image.shape[1]

#we want the new image to be 40% of the original size
scale_factor = 0.4
new_height = int(height * scale_factor)
new_width = int(width * scale_factor)
new_dimensions = (new_width, new_height)
new_image = cv2.resize(image, new_dimensions, interpolation=cv2.INTER_AREA)
print("Original image shape:", image.shape)
print("Resized image shape:", new_image.shape)
cv2.imshow('Resized Image', new_image)
plt.imshow(new_image)
```

### Color Conversion

```python
# 1. Grayscaling the image
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2.imshow('Grayscale Image', gray_image)
plt.imshow(gray_image, cmap='gray')
```

```python
#2. HSV color space
hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
cv2.imshow('HSV Image', hsv_image)
plt.imshow(hsv_image)   
```

## Adjusting the contrast and brightness of image

```python
    def brightness_contrast_adjustment(image, alpha=0, beta=0):
        #alpha: contrast control (1.0-3.0)
        #beta: brightness control (0-100)

        adjusted_image = cv2.convertScaleAbs(image, alpha=alpha, beta=beta)
        return adjusted_image   
        
```

```python
new_image = brightness_contrast_adjustment(image, alpha=1, beta=50)

print("Original image shape")
cv2.imshow('Original Image', image)
plt.imshow(image)

print("/n")

#print("Brightness and Contrast adjusted image shape")
#cv2.imshow('Brightness and Contrast Adjusted Image', new_image)
#plt.imshow(new_image)
```

## Gaussian Blurring

```python
Gaussian1 = cv2.GaussianBlur(image, ksize=(5, 5), sigmaX=0)
Gaussian2 = cv2.GaussianBlur(image, ksize=(11, 11), sigmaX=0)
print("Original image")
cv2.imshow('Original Image', image)
plt.imshow(image)


```

```python
print('\n')

print("Output of Gaussian Blur with kernel size (5, 5)")
cv2.imshow('Gaussian Blur (5, 5)', Gaussian1)
plt.imshow(Gaussian1)
```

```python
print('\n')

print("Output of Gaussian Blur with kernel size (11, 11)")
cv2.imshow('Gaussian Blur (11, 11)', Gaussian2) 
plt.imshow(Gaussian2)
```
