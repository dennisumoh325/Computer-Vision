```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# reading the image
image = cv2.imread(r'C:\\Users\\New\\Desktop\\Jupyter Notebook\\ML\\Computer Vision\\OpenCV for images\\opencv.jpg')

#displaying the image with opencv
cv2.imshow('Image', image)
plt.imshow(image)
```

```python


# Check whether the image loaded
if image is not None:

    # Convert BGR to RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Display image
    plt.imshow(image_rgb)
    plt.title("My Image")
    plt.axis("off")
    plt.show()

else:
    print("Image not found. Check your file path.")
```

### Resizing the image

```python
height = 400
width = 400
dimensions = (width, height)
new_image = cv2.resize(image, dimensions )
print("Original image shape:", image.shape)
print("Resized image shape:", new_image.shape)
cv2.imshow('Resized Image', new_image)
plt.imshow(new_image)
```

```python
# Resizing the image using INTER_AREA interpolation
height = 400
width = 400
dimensions = (width, height)
new_image = cv2.resize(image, dimensions, interpolation=cv2.INTER_AREA) 
print("Original image shape:", image.shape)
print("Resized image shape:", new_image.shape)
cv2.imshow('Resized Image', new_image)
plt.imshow(new_image)
```

```python
# Resizing with scaling factors
height = image.shape[0]
width = image.shape[1]

#we want the new image to be 40% of the original size
scale_factor = 0.4
new_height = int(height * scale_factor)
new_width = int(width * scale_factor)
new_dimensions = (new_width, new_height)
new_image = cv2.resize(image, new_dimensions, interpolation=cv2.INTER_AREA)
print("Original image shape:", image.shape)
print("Resized image shape:", new_image.shape)
cv2.imshow('Resized Image', new_image)
plt.imshow(new_image)
```
