```python
# import the necessary libraries
import PIL
from PIL import Image 
import matplotlib.pyplot as plt
```

```python
#Loading the image
image = Image.open(r'C:\Users\New\Desktop\Jupyter Notebook\ML\Computer Vision\test_image-2.png')
```

```python
#Extracting the pixel map of the image
pixel_values = list(image.getdata())
```

```python
#extracting pixel map
pixel_matrix =image.load()  
```

```python
display(image)
```

```python
#getting the width and height of the image
width, height = image.size
print( width, height)
```

```python
#converting the image to grayscale
for i in range(width):
    for j in range(height):
        r, g, b = pixel_matrix[i, j]
        gray = int(0.2989 * r + 0.5870 * g + 0.1140 * b)
        pixel_matrix[i, j] = (gray, gray, gray)
```
